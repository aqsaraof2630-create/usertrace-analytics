```python
# ============================================================
# USERTRACE ANALYTICS DASHBOARD
# Interactive User Traceability & Usability Analytics Website
# Google Colab - Single Cell
# ============================================================

from IPython.display import HTML, display

html_code = r'''
<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">

<title>UserTrace Analytics</title>

<style>
*{
    margin:0;
    padding:0;
    box-sizing:border-box;
    font-family:Inter,Segoe UI,Arial,sans-serif;
}

body{
    background:#08111f;
    color:#e8f0f7;
    min-height:100vh;
}

/* ================= SIDEBAR ================= */

.sidebar{
    position:fixed;
    left:0;
    top:0;
    width:245px;
    height:100vh;
    background:#0d1a2b;
    border-right:1px solid #1d3448;
    padding:25px 18px;
    z-index:1000;
}

.logo{
    display:flex;
    align-items:center;
    gap:11px;
    margin-bottom:40px;
    padding-left:8px;
}

.logo-icon{
    width:38px;
    height:38px;
    background:#13c9bd;
    color:#06131e;
    border-radius:10px;
    display:flex;
    align-items:center;
    justify-content:center;
    font-weight:800;
    font-size:19px;
}

.logo h2{
    font-size:20px;
    color:white;
}

.logo span{
    color:#13c9bd;
}

.nav-title{
    font-size:11px;
    color:#71879a;
    text-transform:uppercase;
    letter-spacing:1.3px;
    margin:20px 12px 10px;
}

.nav-btn{
    width:100%;
    border:none;
    background:transparent;
    color:#9fb2c3;
    padding:13px 14px;
    margin:3px 0;
    border-radius:9px;
    text-align:left;
    cursor:pointer;
    font-size:14px;
    transition:.2s;
}

.nav-btn:hover,
.nav-btn.active{
    background:#122b3d;
    color:#13c9bd;
}

.nav-btn span{
    margin-right:10px;
}

.session-box{
    position:absolute;
    bottom:25px;
    left:18px;
    right:18px;
    background:#102337;
    border:1px solid #203b51;
    padding:14px;
    border-radius:10px;
}

.session-box small{
    color:#71879a;
}

.session-id{
    font-size:11px;
    color:#13c9bd;
    margin-top:6px;
    word-break:break-all;
}

/* ================= MAIN ================= */

.main{
    margin-left:245px;
    padding:28px 35px;
}

.topbar{
    display:flex;
    justify-content:space-between;
    align-items:center;
    margin-bottom:30px;
}

.topbar h1{
    font-size:27px;
    color:white;
}

.topbar p{
    color:#71879a;
    margin-top:5px;
    font-size:13px;
}

.live{
    display:flex;
    align-items:center;
    gap:8px;
    background:#102a2b;
    border:1px solid #205c5a;
    color:#4ce0d2;
    padding:9px 14px;
    border-radius:20px;
    font-size:12px;
}

.live-dot{
    width:8px;
    height:8px;
    background:#13c9bd;
    border-radius:50%;
    box-shadow:0 0 10px #13c9bd;
}

/* ================= CARDS ================= */

.cards{
    display:grid;
    grid-template-columns:repeat(4,1fr);
    gap:18px;
    margin-bottom:22px;
}

.card{
    background:#0d1a2b;
    border:1px solid #1b3448;
    border-radius:13px;
    padding:20px;
    position:relative;
    overflow:hidden;
}

.card:after{
    content:"";
    position:absolute;
    width:80px;
    height:80px;
    background:#13c9bd;
    opacity:.04;
    right:-20px;
    bottom:-25px;
    border-radius:50%;
}

.card-top{
    display:flex;
    justify-content:space-between;
    color:#8197a9;
    font-size:12px;
}

.card-icon{
    width:35px;
    height:35px;
    background:#112d3c;
    color:#13c9bd;
    border-radius:9px;
    display:flex;
    align-items:center;
    justify-content:center;
}

.card-value{
    font-size:28px;
    font-weight:700;
    color:white;
    margin-top:13px;
}

.card-change{
    font-size:11px;
    color:#4ce0d2;
    margin-top:7px;
}

/* ================= GRID ================= */

.content-grid{
    display:grid;
    grid-template-columns:1.5fr 1fr;
    gap:20px;
    margin-bottom:20px;
}

.panel{
    background:#0d1a2b;
    border:1px solid #1b3448;
    border-radius:13px;
    padding:21px;
}

.panel-header{
    display:flex;
    justify-content:space-between;
    align-items:center;
    margin-bottom:20px;
}

.panel-header h3{
    color:white;
    font-size:16px;
}

.panel-header span{
    font-size:11px;
    color:#71879a;
}

/* ================= CHART ================= */

.chart{
    height:240px;
    display:flex;
    align-items:flex-end;
    gap:13px;
    padding:15px 5px 5px;
    border-bottom:1px solid #1c3448;
}

.bar-wrapper{
    flex:1;
    height:100%;
    display:flex;
    flex-direction:column;
    justify-content:flex-end;
    align-items:center;
}

.bar{
    width:75%;
    max-width:40px;
    background:linear-gradient(to top,#13c9bd,#287f96);
    border-radius:6px 6px 0 0;
    min-height:8px;
    transition:.3s;
}

.bar:hover{
    filter:brightness(1.25);
}

.bar-label{
    color:#71879a;
    font-size:10px;
    margin-top:9px;
}

/* ================= BUTTON ANALYTICS ================= */

.button-list{
    display:flex;
    flex-direction:column;
    gap:12px;
}

.button-row{
    display:grid;
    grid-template-columns:1fr 55px 85px;
    align-items:center;
    gap:10px;
}

.button-name{
    font-size:13px;
    color:#c9d6df;
}

.progress-bg{
    height:7px;
    background:#172a3b;
    border-radius:10px;
    overflow:hidden;
}

.progress{
    height:100%;
    background:#13c9bd;
    border-radius:10px;
    transition:.4s;
}

.click-number{
    text-align:right;
    font-size:12px;
    color:#13c9bd;
}

/* ================= ACTIVITY ================= */

.activity{
    display:flex;
    flex-direction:column;
    gap:0;
}

.activity-row{
    display:grid;
    grid-template-columns:10px 1fr auto;
    gap:12px;
    align-items:center;
    padding:13px 0;
    border-bottom:1px solid #172c3e;
}

.activity-dot{
    width:7px;
    height:7px;
    border-radius:50%;
    background:#13c9bd;
}

.activity-text{
    font-size:12px;
    color:#a8bac8;
}

.activity-text strong{
    color:white;
}

.activity-time{
    color:#647b8e;
    font-size:10px;
}

/* ================= HEATMAP ================= */

.heatmap{
    height:280px;
    position:relative;
    background:
        linear-gradient(#132638 1px,transparent 1px),
        linear-gradient(90deg,#132638 1px,transparent 1px);
    background-size:35px 35px;
    border-radius:9px;
    overflow:hidden;
}

.click-point{
    position:absolute;
    width:14px;
    height:14px;
    background:#ff6b6b;
    border:2px solid white;
    border-radius:50%;
    transform:translate(-50%,-50%);
    box-shadow:0 0 15px rgba(255,107,107,.7);
    animation:pop .25s ease;
}

@keyframes pop{
    from{transform:translate(-50%,-50%) scale(0);}
    to{transform:translate(-50%,-50%) scale(1);}
}

/* ================= DEMO AREA ================= */

.demo-section{
    margin-top:20px;
}

.demo-buttons{
    display:flex;
    flex-wrap:wrap;
    gap:10px;
}

.demo-btn{
    border:none;
    padding:11px 18px;
    border-radius:8px;
    cursor:pointer;
    font-weight:600;
    transition:.2s;
}

.demo-btn.primary{
    background:#13c9bd;
    color:#06131e;
}

.demo-btn.secondary{
    background:#18334a;
    color:#c9d8e3;
}

.demo-btn.warning{
    background:#443525;
    color:#ffcb78;
}

.demo-btn:hover{
    transform:translateY(-2px);
    filter:brightness(1.1);
}

/* ================= TABLE ================= */

.table-panel{
    margin-top:20px;
}

table{
    width:100%;
    border-collapse:collapse;
}

th{
    text-align:left;
    color:#71879a;
    font-size:11px;
    padding:12px;
    border-bottom:1px solid #20384b;
}

td{
    padding:12px;
    font-size:12px;
    color:#b9c9d4;
    border-bottom:1px solid #162b3c;
}

.status{
    color:#4ce0d2;
}

/* ================= MODAL ================= */

.modal{
    display:none;
    position:fixed;
    inset:0;
    background:rgba(0,0,0,.7);
    z-index:2000;
    align-items:center;
    justify-content:center;
}

.modal-box{
    width:430px;
    background:#0d1a2b;
    border:1px solid #29465b;
    border-radius:14px;
    padding:25px;
}

.modal-box h2{
    color:white;
    margin-bottom:12px;
}

.modal-box p{
    color:#91a7b7;
    font-size:13px;
    line-height:1.6;
}

.close-btn{
    margin-top:20px;
    width:100%;
    padding:11px;
    border:none;
    border-radius:8px;
    background:#13c9bd;
    cursor:pointer;
    font-weight:bold;
}

/* ================= RESPONSIVE ================= */

@media(max-width:1000px){
    .cards{
        grid-template-columns:repeat(2,1fr);
    }

    .content-grid{
        grid-template-columns:1fr;
    }
}

@media(max-width:700px){
    .sidebar{
        width:70px;
    }

    .logo h2,
    .nav-title,
    .nav-btn span,
    .session-box{
        display:none;
    }

    .main{
        margin-left:70px;
        padding:20px;
    }

    .cards{
        grid-template-columns:1fr;
    }
}
</style>
</head>

<body>

<!-- ================= SIDEBAR ================= -->

<aside class="sidebar">

    <div class="logo">
        <div class="logo-icon">UT</div>
        <h2>User<span>Trace</span></h2>
    </div>

    <div class="nav-title">Analytics</div>

    <button class="nav-btn active" onclick="showDashboard()">
        <span>▦</span> Dashboard
    </button>

    <button class="nav-btn" onclick="showHeatmap()">
        <span>◉</span> Click Heatmap
    </button>

    <button class="nav-btn" onclick="showActivity()">
        <span>↯</span> Activity Logs
    </button>

    <div class="nav-title">Tools</div>

    <button class="nav-btn" onclick="downloadLogs()">
        <span>↓</span> Export Data
    </button>

    <button class="nav-btn" onclick="clearData()">
        <span>⌫</span> Clear Session
    </button>

    <div class="session-box">
        <small>Current Session</small>
        <div class="session-id" id="sessionDisplay"></div>
    </div>

</aside>


<!-- ================= MAIN ================= -->

<main class="main">

    <div class="topbar">

        <div>
            <h1>UserTrace Analytics</h1>
            <p>User interaction & usability tracking dashboard</p>
        </div>

        <div class="live">
            <div class="live-dot"></div>
            LIVE TRACKING
        </div>

    </div>


    <!-- ================= STAT CARDS ================= -->

    <section class="cards">

        <div class="card">
            <div class="card-top">
                Total Clicks
                <div class="card-icon">⌁</div>
            </div>
            <div class="card-value" id="totalClicks">0</div>
            <div class="card-change">Live interaction count</div>
        </div>

        <div class="card">
            <div class="card-top">
                Active Session
                <div class="card-icon">◉</div>
            </div>
            <div class="card-value">1</div>
            <div class="card-change">Currently tracking</div>
        </div>

        <div class="card">
            <div class="card-top">
                Tracked Elements
                <div class="card-icon">◎</div>
            </div>
            <div class="card-value" id="trackedElements">0</div>
            <div class="card-change">Unique elements clicked</div>
        </div>

        <div class="card">
            <div class="card-top">
                Scroll Depth
                <div class="card-icon">↕</div>
            </div>
            <div class="card-value" id="scrollDepth">0%</div>
            <div class="card-change">Maximum page depth</div>
        </div>

    </section>


    <!-- ================= CHART + BUTTON ANALYTICS ================= -->

    <section class="content-grid">

        <div class="panel">

            <div class="panel-header">
                <h3>Click Activity</h3>
                <span>Last 7 activity groups</span>
            </div>

            <div class="chart" id="clickChart">

                <div class="bar-wrapper">
                    <div class="bar" style="height:20%"></div>
                    <div class="bar-label">1</div>
                </div>

                <div class="bar-wrapper">
                    <div class="bar" style="height:35%"></div>
                    <div class="bar-label">2</div>
                </div>

                <div class="bar-wrapper">
                    <div class="bar" style="height:45%"></div>
                    <div class="bar-label">3</div>
                </div>

                <div class="bar-wrapper">
                    <div class="bar" style="height:25%"></div>
                    <div class="bar-label">4</div>
                </div>

                <div class="bar-wrapper">
                    <div class="bar" style="height:55%"></div>
                    <div class="bar-label">5</div>
                </div>

                <div class="bar-wrapper">
                    <div class="bar" style="height:70%"></div>
                    <div class="bar-label">6</div>
                </div>

                <div class="bar-wrapper">
                    <div class="bar" style="height:85%"></div>
                    <div class="bar-label">7</div>
                </div>

            </div>

        </div>


        <div class="panel">

            <div class="panel-header">
                <h3>Button Performance</h3>
                <span>Click frequency</span>
            </div>

            <div class="button-list" id="buttonAnalytics">

                <div class="button-row">
                    <div class="button-name">Explore Analytics</div>
                    <div class="progress-bg">
                        <div class="progress" id="progress-explore" style="width:0%"></div>
                    </div>
                    <div class="click-number" id="count-explore">0 clicks</div>
                </div>

                <div class="button-row">
                    <div class="button-name">View Report</div>
                    <div class="progress-bg">
                        <div class="progress" id="progress-report" style="width:0%"></div>
                    </div>
                    <div class="click-number" id="count-report">0 clicks</div>
                </div>

                <div class="button-row">
                    <div class="button-name">Start Tracking</div>
                    <div class="progress-bg">
                        <div class="progress" id="progress-track" style="width:0%"></div>
                    </div>
                    <div class="click-number" id="count-track">0 clicks</div>
                </div>

            </div>

        </div>

    </section>


    <!-- ================= DEMO INTERACTION AREA ================= -->

    <section class="panel demo-section">

        <div class="panel-header">
            <h3>Interactive Test Area</h3>
            <span>Click buttons to generate analytics</span>
        </div>

        <div class="demo-buttons">

            <button class="demo-btn primary"
                    id="exploreBtn"
                    onclick="trackDemoButton('explore','Explore Analytics')">
                Explore Analytics
            </button>

            <button class="demo-btn secondary"
                    id="reportBtn"
                    onclick="trackDemoButton('report','View Report')">
                View Report
            </button>

            <button class="demo-btn primary"
                    id="trackBtn"
                    onclick="trackDemoButton('track','Start Tracking')">
                Start Tracking
            </button>

            <button class="demo-btn warning"
                    onclick="openModal()">
                View User Info
            </button>

            <button class="demo-btn secondary"
                    onclick="showNotification('Settings updated successfully!')">
                Settings
            </button>

        </div>

    </section>


    <!-- ================= HEATMAP ================= -->

    <section class="panel" id="heatmapPanel" style="margin-top:20px">

        <div class="panel-header">
            <h3>Live Click Heatmap</h3>
            <span>Click positions recorded in this session</span>
        </div>

        <div class="heatmap" id="heatmap">

            <div style="
                position:absolute;
                top:50%;
                left:50%;
                transform:translate(-50%,-50%);
                color:#4d6577;
                font-size:13px;
                pointer-events:none;">
                Click anywhere on the page
            </div>

        </div>

    </section>


    <!-- ================= ACTIVITY LOG ================= -->

    <section class="panel table-panel">

        <div class="panel-header">
            <h3>Recent User Activity</h3>
            <span id="activityCount">0 events</span>
        </div>

        <table>

            <thead>
                <tr>
                    <th>EVENT</th>
                    <th>ELEMENT</th>
                    <th>TEXT</th>
                    <th>TIME</th>
                    <th>STATUS</th>
                </tr>
            </thead>

            <tbody id="activityTable">

            </tbody>

        </table>

    </section>

</main>


<!-- ================= MODAL ================= -->

<div class="modal" id="userModal">

    <div class="modal-box">

        <h2>User Session Information</h2>

        <p>
            This dashboard demonstrates usability tracking.
            User interactions such as clicks, scroll depth,
            button usage and click coordinates are recorded
            during the current browser session.
        </p>

        <br>

        <p>
            <strong>Session:</strong>
            <span id="modalSession"></span>
        </p>

        <button class="close-btn" onclick="closeModal()">
            Close
        </button>

    </div>

</div>


<script>

/* ============================================================
   USERTRACE ANALYTICS ENGINE
   Based on the provided usability tracker concept
   ============================================================ */

const usabilityLogs = [];

const SESSION_ID =
    'session_' + Math.random().toString(36).substring(2,9);

document.getElementById("sessionDisplay").innerText = SESSION_ID;
document.getElementById("modalSession").innerText = SESSION_ID;


/* ================= COORDINATES ================= */

function getPageCoordinates(event){

    const scrollLeft =
        window.pageXOffset ||
        document.documentElement.scrollLeft;

    const scrollTop =
        window.pageYOffset ||
        document.documentElement.scrollTop;

    return {

        pageX:event.clientX + scrollLeft,

        pageY:event.clientY + scrollTop,

        clientX:event.clientX,

        clientY:event.clientY,

        viewportWidth:window.innerWidth,

        viewportHeight:window.innerHeight
    };
}


/* ================= CLICK TRACKING ================= */

document.addEventListener("click",function(event){

    const coords = getPageCoordinates(event);

    const target = event.target;

    const logEntry = {

        sessionId:SESSION_ID,

        eventType:"click",

        timestamp:new Date().toISOString(),

        timeOffsetMs:performance.now(),

        targetTag:target.tagName,

        targetId:target.id || null,

        targetClass:
            typeof target.className === "string"
            ? target.className
            : null,

        targetText:
            (target.innerText || "")
            .substring(0,40),

        x:coords.pageX,

        y:coords.pageY,

        viewportW:coords.viewportWidth,

        viewportH:coords.viewportHeight
    };

    usabilityLogs.push(logEntry);

    addHeatmapPoint(coords);

    updateDashboard();

    addActivity(logEntry);

},true);


/* ================= MOUSE MOVEMENT ================= */

let lastMoveTime = 0;

document.addEventListener("mousemove",function(event){

    const now = performance.now();

    if(now-lastMoveTime > 250){

        const coords =
            getPageCoordinates(event);

        usabilityLogs.push({

            sessionId:SESSION_ID,

            eventType:"mousemove",

            timestamp:new Date().toISOString(),

            x:coords.pageX,

            y:coords.pageY
        });

        lastMoveTime = now;
    }

});


/* ================= SCROLL TRACKING ================= */

let maxScroll = 0;

window.addEventListener("scroll",function(){

    const scrollTop =
        window.scrollY;

    const pageHeight =
        document.documentElement.scrollHeight -
        window.innerHeight;

    if(pageHeight > 0){

        const percentage =
            Math.round(
                (scrollTop/pageHeight)*100
            );

        maxScroll =
            Math.max(maxScroll,percentage);

        document.getElementById(
            "scrollDepth"
        ).innerText =
            maxScroll + "%";
    }

});


/* ================= HEATMAP ================= */

function addHeatmapPoint(coords){

    const heatmap =
        document.getElementById("heatmap");

    const point =
        document.createElement("div");

    point.className =
        "click-point";

    const rect =
        heatmap.getBoundingClientRect();

    let x =
        coords.clientX - rect.left;

    let y =
        coords.clientY - rect.top;

    /* Keep points inside heatmap */

    x = Math.max(5,Math.min(x,rect.width-5));

    y = Math.max(5,Math.min(y,rect.height-5));

    point.style.left =
        x + "px";

    point.style.top =
        y + "px";

    heatmap.appendChild(point);
}


/* ================= DASHBOARD UPDATE ================= */

function updateDashboard(){

    const clicks =
        usabilityLogs.filter(
            e => e.eventType === "click"
        );

    document.getElementById(
        "totalClicks"
    ).innerText = clicks.length;


    const uniqueElements =
        new Set(
            clicks.map(
                e => e.targetId || e.targetTag
            )
        );

    document.getElementById(
        "trackedElements"
    ).innerText =
        uniqueElements.size;


    updateButtonAnalytics();

    updateChart();
}


/* ================= BUTTON ANALYTICS ================= */

function updateButtonAnalytics(){

    const clicks =
        usabilityLogs.filter(
            e => e.eventType === "click"
        );

    const buttons = {

        explore:
            clicks.filter(
                e => e.targetId === "exploreBtn"
            ).length,

        report:
            clicks.filter(
                e => e.targetId === "reportBtn"
            ).length,

        track:
            clicks.filter(
                e => e.targetId === "trackBtn"
            ).length
    };

    const max =
        Math.max(
            1,
            buttons.explore,
            buttons.report,
            buttons.track
        );


    document.getElementById(
        "count-explore"
    ).innerText =
        buttons.explore + " clicks";

    document.getElementById(
        "count-report"
    ).innerText =
        buttons.report + " clicks";

    document.getElementById(
        "count-track"
    ).innerText =
        buttons.track + " clicks";


    document.getElementById(
        "progress-explore"
    ).style.width =
        (buttons.explore/max*100)+"%";

    document.getElementById(
        "progress-report"
    ).style.width =
        (buttons.report/max*100)+"%";

    document.getElementById(
        "progress-track"
    ).style.width =
        (buttons.track/max*100)+"%";
}


/* ================= CHART ================= */

function updateChart(){

    const clicks =
        usabilityLogs.filter(
            e => e.eventType === "click"
        ).length;

    const bars =
        document.querySelectorAll(".bar");

    bars.forEach((bar,index)=>{

        let height =
            Math.min(
                100,
                20 + clicks*7 + index*5
            );

        bar.style.height =
            height + "%";
    });
}


/* ================= ACTIVITY LOG ================= */

function addActivity(log){

    const table =
        document.getElementById(
            "activityTable"
        );

    const row =
        document.createElement("tr");

    const time =
        new Date(
            log.timestamp
        ).toLocaleTimeString();

    row.innerHTML = `

        <td>${log.eventType}</td>

        <td>${log.targetTag}
            ${log.targetId ? "#"+log.targetId : ""}
        </td>

        <td>${log.targetText || "-"}</td>

        <td>${time}</td>

        <td class="status">Tracked</td>
    `;

    table.prepend(row);

    /* Keep only last 10 rows */

    while(table.children.length > 10){

        table.removeChild(
            table.lastChild
        );
    }

    document.getElementById(
        "activityCount"
    ).innerText =
        usabilityLogs.length + " events";
}


/* ================= DEMO BUTTONS ================= */

function trackDemoButton(type,name){

    showNotification(
        name + " clicked — interaction tracked!"
    );
}


/* ================= MODAL ================= */

function openModal(){

    document.getElementById(
        "userModal"
    ).style.display = "flex";
}

function closeModal(){

    document.getElementById(
        "userModal"
    ).style.display = "none";
}


/* ================= NOTIFICATION ================= */

function showNotification(message){

    const notification =
        document.createElement("div");

    notification.innerText =
        message;

    notification.style.position =
        "fixed";

    notification.style.bottom =
        "25px";

    notification.style.right =
        "25px";

    notification.style.background =
        "#13c9bd";

    notification.style.color =
        "#06131e";

    notification.style.padding =
        "13px 20px";

    notification.style.borderRadius =
        "9px";

    notification.style.fontWeight =
        "600";

    notification.style.zIndex =
        "5000";

    notification.style.boxShadow =
        "0 8px 30px rgba(0,0,0,.35)";

    document.body.appendChild(
        notification
    );

    setTimeout(()=>{

        notification.remove();

    },2000);
}


/* ================= EXPORT JSON ================= */

function downloadLogs(){

    const data =
        JSON.stringify(
            usabilityLogs,
            null,
            2
        );

    const blob =
        new Blob(
            [data],
            {type:"application/json"}
        );

    const url =
        URL.createObjectURL(blob);

    const link =
        document.createElement("a");

    link.href = url;

    link.download =
        "usertrace-session-data.json";

    link.click();

    URL.revokeObjectURL(url);

    showNotification(
        "Analytics data exported successfully!"
    );
}


/* ================= CLEAR SESSION ================= */

function clearData(){

    if(
        !confirm(
            "Clear all current session analytics?"
        )
    ){
        return;
    }

    usabilityLogs.length = 0;

    document.getElementById(
        "totalClicks"
    ).innerText = "0";

    document.getElementById(
        "trackedElements"
    ).innerText = "0";

    document.getElementById(
        "activityTable"
    ).innerHTML = "";

    document.getElementById(
        "activityCount"
    ).innerText = "0 events";

    document.getElementById(
        "heatmap"
    ).querySelectorAll(
        ".click-point"
    ).forEach(
        point => point.remove()
    );

    updateButtonAnalytics();

    showNotification(
        "Session analytics cleared."
    );
}


/* ================= NAVIGATION ================= */

function showDashboard(){

    window.scrollTo({
        top:0,
        behavior:"smooth"
    });

    showNotification(
        "Dashboard view opened"
    );
}

function showHeatmap(){

    document.getElementById(
        "heatmapPanel"
    ).scrollIntoView({
        behavior:"smooth"
    });

    showNotification(
        "Click heatmap opened"
    );
}

function showActivity(){

    document.querySelector(
        ".table-panel"
    ).scrollIntoView({
        behavior:"smooth"
    });

    showNotification(
        "Activity logs opened"
    );
}


/* ================= PUBLIC TRACKER API ================= */

window.UserTraceTracker = {

    getLogs:function(){

        return usabilityLogs;
    },

    exportJSON:function(){

        return JSON.stringify(
            usabilityLogs,
            null,
            2
        );
    },

    getSessionID:function(){

        return SESSION_ID;
    },

    getTotalClicks:function(){

        return usabilityLogs.filter(
            e => e.eventType === "click"
        ).length;
    }
};


/* Initial dashboard */

updateDashboard();

</script>

</body>
</html>
'''

display(HTML(html_code))
```
