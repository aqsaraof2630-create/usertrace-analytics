```python

from IPython.display import HTML, display

html = r"""
<style>
*{box-sizing:border-box;font-family:'Segoe UI',sans-serif}
body{background:#0b1020}
.app{max-width:850px;margin:auto;padding:24px;background:#0e172b;color:#f1f5f9;border-radius:24px;border:1px solid #293650}
header{display:flex;justify-content:space-between;align-items:center;gap:12px;flex-wrap:wrap}
h1{margin:0;color:#a78bfa;font-size:30px}
.sub{color:#94a3b8;font-size:13px;margin-top:6px}
.status{color:#5eead4;font-size:12px}
.hero{margin:24px 0;padding:24px;border-radius:20px;background:linear-gradient(130deg,#29234c,#123b43);display:flex;align-items:center;justify-content:space-between;gap:20px;flex-wrap:wrap}
.hero h2{font-size:27px;margin:8px 0}
.hero p{color:#c1cce0;line-height:1.6;font-size:14px;max-width:400px}
.ring{width:155px;height:155px;position:relative;flex-shrink:0}
.ring svg{width:100%;height:100%;transform:rotate(-90deg)}
.center{position:absolute;inset:0;display:flex;align-items:center;justify-content:center;flex-direction:column}
.num{font-size:45px;font-weight:800}
.muted{color:#94a3b8;font-size:12px}
.modes{display:grid;grid-template-columns:repeat(3,1fr);gap:10px;margin:15px 0 24px}
.mode{background:#18243a;border:1px solid #34415d;border-radius:14px;padding:14px 5px;color:#e2e8f0;cursor:pointer}
.mode strong,.mode small{display:block;margin-top:6px}
.mode small{font-size:11px;color:#94a3b8}
.mode.on{background:#30264f;border-color:#a78bfa}
.grid{display:grid;grid-template-columns:1fr 1fr;gap:16px}
.card{padding:20px;border:1px solid #293650;background:#131e32;border-radius:18px;min-width:0}
h3{margin:0 0 18px;font-size:17px}
.buttons{display:grid;grid-template-columns:1fr 1fr;gap:10px}
.btn{padding:14px 8px;border:0;border-radius:12px;font-weight:700;cursor:pointer;transition:.2s}
.btn:hover{filter:brightness(1.1)}
.btn:active{transform:scale(.97)}
.btn:disabled{opacity:.35;cursor:not-allowed}
.add{background:#5eead4;color:#08352f}
.remove{background:#fcd7a1;color:#54320a}
.undo{background:#a78bfa;color:#27134b}
.reset{background:#fb7185;color:#4c1020}
.goalrow{display:flex;justify-content:space-between;align-items:center;gap:8px;margin-top:18px;padding-top:15px;border-top:1px solid #293650;color:#94a3b8;font-size:13px}
.goalrow button{background:#25334b;color:white;border:0;border-radius:8px;width:30px;height:30px;cursor:pointer}
.goalrow b{color:#5eead4;margin:0 8px}
#message{color:#5eead4;min-height:22px;margin:15px 0;font-size:13px;text-align:center}
#log{max-height:250px;overflow:auto}
.entry{display:flex;justify-content:space-between;gap:8px;border-bottom:1px solid #293650;padding:12px 0;font-size:12px}
.entry span:last-child{color:#94a3b8}
.empty{text-align:center;color:#94a3b8;padding:25px 0;font-size:13px}
.help{color:#94a3b8;font-size:12px;line-height:1.8;margin-top:15px}
video{width:100%;max-width:300px;border-radius:12px;margin-top:12px;transform:scaleX(-1)}
#camera{display:none}
footer{text-align:center;color:#94a3b8;font-size:12px;margin-top:22px}
@media(max-width:550px){.grid{grid-template-columns:1fr}.hero{justify-content:center}.modes{gap:5px}.mode{font-size:12px}.card{padding:15px}}
</style>

<div class="app">
<header>
 <div>
  <h1>✦ FocusFlow</h1>
  <div class="sub">Your personal study companion</div>
 </div>
 <div class="status" id="status">● Ready to focus</div>
</header>

<div class="hero">
 <div>
  <div style="color:#5eead4;font-size:12px;letter-spacing:2px">YOUR DAILY PROGRESS</div>
  <h2>Small steps.<br>Big progress.</h2>
  <p>Complete your study sessions, track your progress and control your tracker using mouse clicks, voice commands or hand gestures.</p>
 </div>
 <div style="text-align:center">
  <div class="ring">
   <svg viewBox="0 0 180 180">
    <circle cx="90" cy="90" r="80" fill="none" stroke="#293650" stroke-width="12"/>
    <circle id="arc" cx="90" cy="90" r="80" fill="none" stroke="#5eead4" stroke-width="12" stroke-linecap="round" stroke-dasharray="502.65" stroke-dashoffset="502.65"/>
   </svg>
   <div class="center">
    <div class="num" id="count">0</div>
    <div class="muted">SESSIONS</div>
   </div>
  </div>
  <div class="muted" id="goalText">Daily goal: 4 sessions</div>
 </div>
</div>

<h3>Choose your control</h3>
<div class="modes">
 <button class="mode on" id="mMouse">🖱️<strong>Mouse</strong><small>Click buttons</small></button>
 <button class="mode" id="mVoice">🎙️<strong>Voice</strong><small>Speak commands</small></button>
 <button class="mode" id="mGest">✋<strong>Gestures</strong><small>Use your hand</small></button>
</div>

<div class="grid">
 <section class="card">
  <h3>Session Controls</h3>
  <div class="buttons">
   <button class="btn add action" id="bAdd">＋ Complete Session</button>
   <button class="btn remove action" id="bSub">－ Remove Session</button>
   <button class="btn undo action" id="bUndo">↶ Undo Action</button>
   <button class="btn reset action" id="bReset">↻ Reset Today</button>
  </div>
  <div class="goalrow">
   <span>Daily goal</span>
   <div>
    <button id="gMinus">−</button>
    <b id="goalNum">4</b>
    <button id="gPlus">+</button>
   </div>
  </div>
  <div id="message" role="status">Ready when you are!</div>
  <div id="camera">
   <div class="help">1 finger = Add · 2 = Remove · 3 = Undo · 4 = Reset</div>
   <video id="cam" autoplay playsinline muted></video>
  </div>
  <div class="help">
   <b style="color:#7dd3fc">Voice commands:</b><br>
   Say "add", "remove", "undo" or "reset".<br>
   <b style="color:#7dd3fc">Gesture commands:</b><br>
   Show 1–4 fingers to the camera.
  </div>
 </section>

 <section class="card">
  <h3>Activity History <span class="muted" id="eventCount">0 events</span></h3>
  <div id="log"><div class="empty">Your activity will appear here.</div></div>
  <div class="help">Every action is recorded with its time. Undo restores your previous session count.</div>
 </section>
</div>
<footer>FOCUSFLOW · Stay consistent, one session at a time ✦</footer>
</div>

<script>
(function(){
 const $=id=>document.getElementById(id);
 let count=0,goal=4,history=[],logs=[],mouseOn=true;
 let voiceOn=false,rec=null,gestOn=false,stream=null,hands=null;
 let loaded=false,lastGesture=-1,stable=0,lastFire=0,busy=false,loopId=null;
 const circumference=2*Math.PI*80;

 function notify(msg){
  $('message').textContent=msg;
  $('status').textContent='● '+msg;
 }

 function stamp(action){
  logs.push({action,time:new Date().toLocaleTimeString()});
 }

 function render(){
  $('count').textContent=count;
  $('goalNum').textContent=goal;
  $('goalText').textContent='Daily goal: '+goal+' sessions';
  const p=Math.min(count/goal,1);
  $('arc').style.strokeDashoffset=circumference*(1-p);
  $('arc').style.stroke=p>=1?'#5eead4':'#a78bfa';
  $('eventCount').textContent=logs.length+' events';
  $('log').replaceChildren();

  if(!logs.length){
   const e=document.createElement('div');
   e.className='empty';
   e.textContent='Your activity will appear here.';
   $('log').appendChild(e);
  }else{
   logs.slice().reverse().forEach(item=>{
    const row=document.createElement('div');
    row.className='entry';
    const a=document.createElement('span');
    const t=document.createElement('span');
    a.textContent=item.action;
    t.textContent=item.time;
    row.append(a,t);
    $('log').appendChild(row);
   });
  }

  $('bSub').disabled=count===0;
  $('bUndo').disabled=history.length===0;
  document.querySelectorAll('.action').forEach(b=>b.disabled=!mouseOn);
  $('gMinus').disabled=goal<=1;
  $('gPlus').disabled=goal>=20;
 }

 const actions={
  add(){
   history.push(count);count++;stamp('+1 study session');render();
   notify(count>=goal?'🎉 Daily goal completed!':'✓ Study session completed!');
  },
  sub(){
   if(count<=0){notify('No sessions to remove.');return;}
   history.push(count);count--;stamp('-1 study session');render();notify('Session removed.');
  },
  undo(){
   if(!history.length){notify('Nothing to undo.');return;}
   count=history.pop();stamp('Undo action');render();notify('Previous count restored.');
  },
  reset(){
   if(count===0){notify('Count is already zero.');return;}
   history.push(count);count=0;stamp('Reset session count');render();notify('Session count reset.');
  }
 };

 $('bAdd').onclick=actions.add;
 $('bSub').onclick=actions.sub;
 $('bUndo').onclick=actions.undo;
 $('bReset').onclick=actions.reset;

 $('gPlus').onclick=()=>{goal=Math.min(20,goal+1);render();notify('Daily goal increased.');};
 $('gMinus').onclick=()=>{goal=Math.max(1,goal-1);render();notify('Daily goal decreased.');};

 $('mMouse').onclick=()=>{
  mouseOn=!mouseOn;
  $('mMouse').classList.toggle('on',mouseOn);
  render();notify(mouseOn?'Mouse controls enabled.':'Mouse controls paused.');
 };

 function handleVoice(text){
  const t=text.toLowerCase().trim();
  if(/\b(reset|clear|start over)\b/.test(t))actions.reset();
  else if(/\b(undo|go back)\b/.test(t))actions.undo();
  else if(/\b(remove|subtract|decrease|minus)\b/.test(t))actions.sub();
  else if(/\b(add|complete|completed|increase|plus)\b/.test(t))actions.add();
  else notify('Command not recognized: '+text);
 }

 $('mVoice').onclick=()=>{
  const SR=window.SpeechRecognition||window.webkitSpeechRecognition;
  if(!SR){notify('Voice recognition unavailable in this browser.');return;}

  if(voiceOn){
   voiceOn=false;if(rec)rec.stop();
   $('mVoice').classList.remove('on');notify('Voice controls disabled.');return;
  }

  rec=new SR();rec.lang='en-US';rec.continuous=true;rec.interimResults=false;
  rec.onresult=e=>handleVoice(e.results[e.results.length-1][0].transcript);
  rec.onerror=e=>{
   if(e.error==='not-allowed'){
    voiceOn=false;$('mVoice').classList.remove('on');
    notify('Please allow microphone permission.');
   }else notify('Voice error: '+e.error);
  };
  rec.onend=()=>{if(voiceOn)try{rec.start();}catch(e){}};

  try{
   rec.start();voiceOn=true;$('mVoice').classList.add('on');
   notify('Listening. Say a command.');
  }catch(e){notify('Could not start voice recognition.');}
 };

 function loadScript(src){
  return new Promise((resolve,reject)=>{
   const s=document.createElement('script');
   s.src=src;s.crossOrigin='anonymous';
   s.onload=resolve;s.onerror=reject;document.head.appendChild(s);
  });
 }

 function countFingers(lm){
  let n=0;
  [[8,6],[12,10],[16,14],[20,18]].forEach(([tip,joint])=>{
   if(lm[tip].y<lm[joint].y)n++;
  });
  return n;
 }

 function onHands(r){
  if(!gestOn||!r.multiHandLandmarks||!r.multiHandLandmarks.length){
   lastGesture=-1;stable=0;return;
  }
  const g=countFingers(r.multiHandLandmarks[0]);
  if(g===lastGesture)stable++;
  else{lastGesture=g;stable=1;}

  if(stable>=(g===4?18:8)&&Date.now()-lastFire>1600&&g>=1&&g<=4){
   lastFire=Date.now();stable=0;
   notify(g+' finger gesture detected.');
   ({1:actions.add,2:actions.sub,3:actions.undo,4:actions.reset})[g]();
  }
 }

 async function cameraLoop(){
  if(!gestOn||busy)return;
  const v=$('cam');
  if(v.readyState>=2){
   busy=true;
   try{await hands.send({image:v});}catch(e){console.error(e);}
   busy=false;
  }
  if(gestOn)loopId=requestAnimationFrame(cameraLoop);
 }

 async function stopCamera(){
  gestOn=false;
  if(loopId!==null)cancelAnimationFrame(loopId);
  loopId=null;
  if(stream){stream.getTracks().forEach(t=>t.stop());stream=null;}
  $('cam').srcObject=null;
  $('camera').style.display='none';
 }

 $('mGest').onclick=async()=>{
  if(gestOn){
   await stopCamera();$('mGest').classList.remove('on');
   notify('Gesture controls disabled.');return;
  }
  try{
   if(!navigator.mediaDevices?.getUserMedia)
    throw new Error('Camera requires supported secure browser access.');

   notify('Requesting camera access...');
   stream=await navigator.mediaDevices.getUserMedia({video:true,audio:false});
   $('cam').srcObject=stream;
   $('camera').style.display='block';
   await $('cam').play();

   if(!loaded){
    await loadScript('https://cdn.jsdelivr.net/npm/@mediapipe/hands@0.4.1675469240/hands.js');
    hands=new Hands({locateFile:f=>'https://cdn.jsdelivr.net/npm/@mediapipe/hands@0.4.1675469240/'+f});
    hands.setOptions({maxNumHands:1,modelComplexity:0,minDetectionConfidence:0.7,minTrackingConfidence:0.6});
    hands.onResults(onHands);loaded=true;
   }

   gestOn=true;$('mGest').classList.add('on');
   notify('Show your hand to the camera.');cameraLoop();
  }catch(e){
   await stopCamera();$('mGest').classList.remove('on');
   notify('Camera unavailable: '+e.message);
  }
 };

 window.addEventListener('pagehide',()=>{
  if(rec)rec.stop();
  if(stream)stream.getTracks().forEach(t=>t.stop());
 });

 render();
})();
</script>
"""

display(HTML(html))
```
